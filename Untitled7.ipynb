```python
def validate_car_number(car_id):
    valid_letters = 'АВЕКМНОРСТУХ'

    if len(car_id) not in [8, 9]:
        return "Номер не валиден."

    if len(car_id) == 8:
        pattern = [
            car_id[0] in valid_letters,
            car_id[1:4].isdigit(),
            car_id[4] in valid_letters,
            car_id[5] in valid_letters,
            car_id[6:8].isdigit()
        ]
        region = car_id[6:8]
        number_part = car_id[:6]
    else:
        pattern = [
            car_id[0] in valid_letters,
            car_id[1:4].isdigit(),
            car_id[4] in valid_letters,
            car_id[5] in valid_letters,
            car_id[6:9].isdigit()
        ]
        region = car_id[6:9]
        number_part = car_id[:6]

    if all(pattern):
        return f"Номер {number_part} валиден. Регион: {region}."
    else:
        return "Номер не валиден."

car_id = 'А222ВС96'
print(validate_car_number(car_id))

car_id = 'А123ВС777'
print(validate_car_number(car_id))

car_id = 'АБ22ВВ193'
print(validate_car_number(car_id))
```
