```python
import pandas as pd, matplotlib.pyplot as plt; file_name = list(uploaded.keys())[0]; df = pd.read_csv(file_name); df['Purchase Date'] = pd.to_datetime(df['Purchase Date']); df['Month'] = df['Purchase Date'].dt.to_period('M'); df['Quarter'] = df['Purchase Date'].dt.to_period('Q'); delivery = df.groupby('Shipping Type')['Total Price'].sum(); products = df.groupby('Product Type')['Total Price'].sum(); monthly = df.groupby('Month')['Add-on Total'].sum(); quarterly = df.groupby('Quarter')['Add-on Total'].sum(); fig, axes = plt.subplots(2, 2, figsize=(15, 10)); axes[0,0].bar(delivery.index, delivery.values); axes[0,0].set_title('Доход по доставке'); axes[0,1].bar(products.index, products.values); axes[0,1].set_title('Доход по продуктам'); axes[1,0].plot(monthly.index.astype(str), monthly.values, marker='o'); axes[1,0].set_title('Допы по месяцам'); axes[1,1].bar(quarterly.index.astype(str), quarterly.values); axes[1,1].set_title('Допы по кварталам'); [ax.tick_params(axis='x', rotation=45) for ax in axes.flat]; plt.tight_layout(); plt.show(); print("Доставка:", delivery.to_dict()); print("Продукты:", products.to_dict()); print("Месяцы:", monthly.to_dict()); print("Кварталы:", quarterly.to_dict())
```
